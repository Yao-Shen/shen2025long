```python
import numpy as np
import matplotlib.pyplot as plt
from plot import make_theme

%matplotlib inline
plt.rcParams['font.size'] = 7
```

```python
def plot_label(ax, count, x=0, y=1.03, color='k'):
    ax.text(x, y, count, fontdict=dict(fontsize=9), transform=ax.transAxes, color=color, weight='bold')
```

```python
fig = plt.figure(figsize=(6.75,6.75/1.5), dpi=150)
gs1 = fig.add_gridspec(nrows=1, ncols=2, left=0.06, right=0.48, top=0.96, bottom=0.56, wspace=0, hspace=0)
gs2 = fig.add_gridspec(nrows=1, ncols=2, left=0.56, right=0.98, top=0.96, bottom=0.56, wspace=0, hspace=0)
gs3 = fig.add_gridspec(nrows=1, ncols=2, left=0.06, right=0.48, top=0.47, bottom=0.07, wspace=0, hspace=0)
gs4 = fig.add_gridspec(nrows=1, ncols=2, left=0.56, right=0.98, top=0.47, bottom=0.07, wspace=0, hspace=0)
axs1 = [fig.add_subplot(gs) for gs in gs1]
axs2 = [fig.add_subplot(gs) for gs in gs2]
axs3 = [fig.add_subplot(gs) for gs in gs3]
axs4 = [fig.add_subplot(gs) for gs in gs4]

δs = np.arange(0.,0.02001,0.0001)
Hs = np.arange(0.,1.01,0.01)

# Mn
for δ in δs:
    evals = np.load(r'../ED_results/Mn_delta_{:.4f}.npy'.format(δ))
    evals = np.unique(np.round(evals, 6))[:20]
    axs1[0].plot(np.ones_like(evals)*δ*1e3, evals*1e3, '.', ms=3, color='C4')
for H in Hs:
    evals = np.load(r'../ED_results/Mn_delta_0.02_Hx_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs1[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=3, color='C4')
for H in Hs:
    evals = np.load(r'../ED_results/Mn_delta_0.02_Hz_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs1[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=1, color='k')

# Fe
for δ in δs:
    evals = np.load(r'../ED_results/Fe_delta_{:.4f}.npy'.format(δ))
    evals = np.unique(np.round(evals, 6))[:20]
    axs2[0].plot(np.ones_like(evals)*δ*1e3, evals*1e3, '.', ms=3, color='C3')
for H in Hs:
    evals = np.load(r'../ED_results/Fe_delta_0.02_Hx_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs2[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=3, color='C3')
for H in Hs:
    evals = np.load(r'../ED_results/Fe_delta_0.02_Hz_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs2[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=1, color='k')

# Co
for δ in δs:
    evals = np.load(r'../ED_results/Co_delta_{:.4f}.npy'.format(δ))
    evals = np.unique(np.round(evals, 6))[:20]
    axs3[0].plot(np.ones_like(evals)*δ*1e3, evals*1e3, '.', ms=3, color='royalblue')
for H in Hs:
    evals = np.load(r'../ED_results/Co_delta_0.02_Hx_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs3[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=3, color='royalblue')
for H in Hs:
    evals = np.load(r'../ED_results/Co_delta_0.02_Hz_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs3[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=1, color='k')

# Ni
for δ in δs:
    evals = np.load(r'../ED_results/Ni_delta_{:.4f}.npy'.format(δ))
    evals = np.unique(np.round(evals, 6))[:20]
    axs4[0].plot(np.ones_like(evals)*δ*1e3, evals*1e3, '.', ms=3, color='C2')
for H in Hs:
    evals = np.load(r'../ED_results/Ni_delta_0.02_Hx_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs4[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=3, color='C2')
for H in Hs:
    evals = np.load(r'../ED_results/Ni_delta_0.02_Hz_{:.2f}.npy'.format(H))
    evals = np.unique(np.round(evals, 6))[:20]
    axs4[1].plot(np.ones_like(evals)*H, evals*1e3, '.', ms=1, color='k')
    
# config
for axs in [axs1,axs2,axs3,axs4]:
    for ax in axs:
        make_theme(ax, ticklength=4)
        ax.set_ylim(-5,50)
    axs[1].tick_params(labelleft=False)
    axs[0].set_xlim(-2.,20.)
    axs[0].set_xticks(np.arange(0.,20.1,8.))
    axs[0].set_xticklabels(['0','-8','-16'])
    axs[1].set_xlim(0.,1.)
    axs[1].set_xticks(np.arange(0.,1.1,0.4))
    axs[1].set_xticklabels(['0','0.4','0.8'])
axs1[0].set_ylabel('Energy (meV)')
axs3[0].set_ylabel('Energy (meV)')
axs3[0].set_xlabel('$\\delta$ (meV)')
axs3[1].set_xlabel('$H$ (meV)')
axs4[0].set_xlabel('$\\delta$ (meV)')
axs4[1].set_xlabel('$H$ (meV)')

# fig.savefig('ED.pdf')
```
