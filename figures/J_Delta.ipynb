```python
import numpy as np
import matplotlib.pyplot as plt
from plot import make_theme

%matplotlib inline
plt.rcParams['font.size'] = 7
```

```python
fig, ax = plt.subplots(figsize=(6.75/2,6.75/3), dpi=150, gridspec_kw=dict(left=0.15, right=0.95, top=0.95, bottom=0.15))

data = dict(np.load('../others/J_vs_Delta.npz'))
ax.plot(data['Js'], data['Theta1'], color='C3', label='Spin 1')
ax.plot(data['Js'], data['Theta2'], color='royalblue', label='Spin 2')
# ax.legend()
make_theme(ax, ticklength=4)
ax.set_xlim(-0.25,2.25)
ax.set_xlabel('$J/\\Delta_1$')
ax.set_ylim(-10,100)
ax.set_ylabel('$\\Theta$ (deg)')

# fig.savefig('J_vs_Delta.pdf')
```
