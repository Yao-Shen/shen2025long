```python
import numpy as np
import matplotlib.pyplot as plt
from plot import make_theme, unicorn, genlattice

%matplotlib inline
plt.rcParams['font.size'] = 7
```

```python
def plot_label(ax, count, x=0, y=1.03, color='k'):
    ax.text(x, y, count, fontdict=dict(fontsize=9), transform=ax.transAxes, color=color, weight='bold')
```

```python
r = unicorn(853.6e3, 'meV')['wavevector']*2 / (genlattice([5.9321, 10.2737, 6.7061, 90, 108, 90])['as']/2) * 0.5
```

```python
b = np.sqrt(3)
c = 2./np.sqrt(3)

fig, axs = plt.subplots(1, 2, figsize=(6.75,6.75/2), dpi=150, gridspec_kw=dict(left=0.05, right=0.95, bottom=0.05, top=0.9))

xs = np.arange(0.,1.01,0.5)
ys = np.array([-0.1674,-0.3326,-0.1674]) * b
axs[0].plot(xs, ys, 'o', color='gray', zorder=1)
axs[0].plot(xs, ys, '-', lw=1, color='darkgray', zorder=-1)
xs = np.arange(-0.5,1.51,0.5)
ys = np.array([0.3326,0.1674,0.3326,0.1674,0.3326]) * b
axs[0].plot(xs, ys, 'o', color='gray', zorder=1)
axs[0].plot(xs, ys, '-', lw=1, color='darkgray', zorder=-1)
xs = np.arange(-0.5,1.51,0.5)
ys = np.array([0.6674,0.8326,0.6674,0.8326,0.6674]) * b
axs[0].plot(xs, ys, 'o', color='gray', zorder=1)
axs[0].plot(xs, ys, '-', lw=1, color='darkgray', zorder=-1)
xs = np.arange(0.,1.01,0.5)
ys = np.array([1.1674,1.3326,1.1674]) * b
axs[0].plot(xs, ys, 'o', color='gray', zorder=1)
axs[0].plot(xs, ys, '-', lw=1, color='darkgray', zorder=-1)
xs = [0.,1.,-0.5,0.5,1.5,0.,1.]
ys = [[-0.1674,0.1674], [-0.1674,0.1674], [0.3326,0.6674], [0.3326,0.6674], [0.3326,0.6674], [0.8326,1.1674], [0.8326,1.1674]]
for x, y in zip(xs, ys):
    axs[0].plot([x,x], np.array(y)*b, '-', lw=1, color='darkgray', zorder=-1)
axs[0].plot([0.,1.,1.,0.,0.], np.array([0.,0.,1.,1.,0.])*b, '--', lw=1, color='C3')
axs[0].plot([0.,0.5,1.,0.5,0.], np.array([0.8326,0.3326,0.8326,1.3326,0.8326])*b, '--', lw=1, color='royalblue')

axs[0].set_aspect('equal')
axs[0].set_xlim(-1.2,1.8)
axs[0].set_ylim(-0.5*b, 1.5*b)
axs[0].axis('off')

arrow0 = [-0.8,-0.2]
arrow_params = dict(head_width=0.1, head_length=0.15, overhang=0.3, lw=0.5, length_includes_head=True)
axs[0].arrow(arrow0[0], arrow0[1], 0.5, 0., color='C3', **arrow_params)
axs[0].arrow(arrow0[0], arrow0[1], 0., 0.5*b, color='C3', **arrow_params)
axs[0].arrow(arrow0[0], arrow0[1], 0.25, -0.25*b, color='royalblue', **arrow_params)
axs[0].arrow(arrow0[0], arrow0[1], 0.25, 0.25*b, color='royalblue', **arrow_params)
axs[0].text(-0.3, -0.16, 'a', color='C3', fontstyle='italic')
axs[0].text(-0.92, 0.7, 'b', color='C3', fontstyle='italic')
axs[0].text(-0.52, -0.6, 'a', color='royalblue', fontstyle='italic')
axs[0].text(-0.52, 0.2, 'b', color='royalblue', fontstyle='italic')
axs[0].text(-1, 2.2, 'Monoclinic', color='C3')
axs[0].text(-1, 2.05, 'Hexagonal', color='royalblue')
axs[0].text(1.1, 2, 'TM', color='gray')

axs[1].plot(0., 1/b, 'D', color='gray', zorder=-10)
axs[1].plot([0.5,-0.5,-0.5,0.5,0.5], np.array([-0.5,-0.5,0.5,0.5,-0.5])/b, '--', lw=1, color='C3')
axs[1].plot([-0.5,-0.5,0.5,0.5], np.array([0.5,1.5,1.5,0.5])/b, '--', lw=1, color='C3')
axs[1].plot(np.array([-0.5,0.5,1.,0.5,-0.5,-1.,-0.5])/np.sqrt(3)*c, np.array([-0.5,-0.5,0.,0.5,0.5,0.,-0.5])*c, '--', lw=1, color='royalblue')

r = unicorn(853.6e3, 'meV')['wavevector']*2 / (genlattice([5.9321, 10.2737, 6.7061, 90, 108, 90])['as']/2) * 0.5
angles = np.arange(0.,360.1,1.)
axs[1].plot(r*np.cos(np.deg2rad(angles)), r*np.sin(np.deg2rad(angles)), color='C2', lw=0.5)

axs[1].set_aspect('equal')
axs[1].set_xlim(-1.2,0.8)
axs[1].axis('off')

arrow0 = [-1.05,-0.7]
arrow_params = dict(head_width=0.06667, head_length=0.1, overhang=0.3, lw=0.5, length_includes_head=True)
axs[1].arrow(arrow0[0], arrow0[1], 0.5, 0., color='C3', **arrow_params)
axs[1].arrow(arrow0[0], arrow0[1], 0., 0.5/b, color='C3', **arrow_params)
axs[1].arrow(arrow0[0], arrow0[1], 0.5, -0.5/np.sqrt(3), color='royalblue', **arrow_params)
axs[1].arrow(arrow0[0], arrow0[1], 0.5, 0.5/np.sqrt(3), color='royalblue', **arrow_params)
axs[1].text(-0.55, -0.67, 'a*', color='C3', fontstyle='italic')
axs[1].text(-1.03, -0.43, 'b*', color='C3', fontstyle='italic')
axs[1].text(-0.55, -0.96, 'a*', color='royalblue', fontstyle='italic')
axs[1].text(-0.55, -0.52, 'b*', color='royalblue', fontstyle='italic')

axs[0].set_title('Real space', fontsize=7)
axs[1].set_title('Reciprocal space', fontsize=7)

plot_label(axs[0], 'a', x=0, y=1.03, color='k')
plot_label(axs[1], 'b', x=0, y=1.03, color='k')

# fig.savefig('unit_cell.pdf')
```
